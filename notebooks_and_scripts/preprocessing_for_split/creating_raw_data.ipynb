#### Preparation of endpoint-specific datasets for split generation
This script prepares MAOMAO data for the dataset splitting workflow. It creates one binary classification dataset for each toxicity endpoint while preserving the original sequence identifiers.

Overview
- Task: Preparation of endpoint-specific datasets for splitting
- Input: MAOMAO sequence-level pivot dataset
- Output: One sequences.csv file per toxicity endpoint

Process
- Load maomao_sequence_pivot.csv
- Iterate over the toxicity endpoint columns
- Preserve the original id and sequence columns
- Rename the selected endpoint column to label
- Retain only positive (1) and negative (0) annotations
- Exclude ambiguous, unlabeled, and missing-information records
- Export each binary dataset to:

- Auxiliary variables

```python
path_data = "../../processed_data"
path_export = "../../pipelines/data"
```

- Read data

```python
df_data = pd.read_csv(f"{path_data}/processed_data/maomao_sequence_pivot.csv")
df_data.head()
```

```python
for column in df_data.columns:
    if ((column != "sequence") & (column != "id")):
        print("Processing column:", column)

        # Select columns
        df = df_data[["id", "sequence", column]].copy()
        df.columns = ["id", "sequence", "label"]

        # filter labeled
        df_labeled = df[(df["label"] == 1) | (df["label"] == 0)].copy() # Select only positive and negative values ​​for split
        df_labeled = df_labeled[["id", "sequence", "label"]]

        os.makedirs(f"{path_export}/{column}", exist_ok=True)
        df_labeled.to_csv(f"{path_export}/{column}/sequences.csv", index=False)
```
